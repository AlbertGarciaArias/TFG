```python
import os
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt

def generate_chain_cluster(start, n_proteins, link_min, link_max, overlap_min, max_tries=50):
    proteins = [np.array(start, dtype=float)]

    for _ in range(1, n_proteins):
        placed = False

        for _ in range(max_tries):
            angle = np.random.uniform(0, 2 * np.pi)
            step = np.random.uniform(link_min, link_max)
            candidate = proteins[-1] + step * np.array([np.cos(angle), np.sin(angle)])

            if all(np.linalg.norm(candidate - p) >= overlap_min for p in proteins):
                proteins.append(candidate)
                placed = True
                break

        if not placed:
            break

    proteins = np.array(proteins)
    center = np.mean(proteins, axis=0)
    return proteins, center


# ============================
# CONFIGURACIÓ
# ============================

H = 1000
W = H
lim = 100

number_of_clusters = 7
min_number_of_proteins = 7
max_number_of_proteins = 14
mean_number_of_proteins = (min_number_of_proteins + max_number_of_proteins) / 2

link_min = 20
link_max = 60
overlap_min = link_min

p_geo = 1/5
stdev_loc_prec = 15
noise_level = 1 * number_of_clusters * mean_number_of_proteins * (1/p_geo)

disp = False

output_folder = "test_dataset"
os.makedirs(output_folder, exist_ok=True)

# ============================
# GENERAR 50 DOCUMENTS
# ============================

for run in range(50):

    data = []   # 👈 IMPORTANT: reiniciar per cada document

    seed_coord = np.random.uniform(lim, H-lim, size=(number_of_clusters, 2))
    cluster_coord = np.zeros((number_of_clusters, 2))
    noise_coord = np.random.uniform(0, H, size=(int(noise_level), 2))

    index_cluster = 0

    if disp:
        fig = plt.figure(figsize=(10, 10))

    for i in range(number_of_clusters):
        index_protein = 0
        index_cluster += 1

        number_of_proteins = np.random.randint(min_number_of_proteins, max_number_of_proteins + 1)

        protein_coord, cc = generate_chain_cluster(
            seed_coord[i],
            number_of_proteins,
            link_min,
            link_max,
            overlap_min
        )

        cluster_coord[i, :] = cc

        for x, y in protein_coord:
            index_protein += 1

            number_of_locs = np.random.geometric(p=p_geo)

            for k in range(number_of_locs):
                x_loc = x + np.random.normal(0, stdev_loc_prec)
                y_loc = y + np.random.normal(0, stdev_loc_prec)

                if 0 < x_loc < H and 0 < y_loc < H:
                    data.append([x_loc/H, y_loc/H, index_cluster, index_protein, run])

    # Afegir soroll
    for i in range(len(noise_coord)):
        x_noise, y_noise = noise_coord[i]
        if 0 < x_noise < H and 0 < y_noise < H:
            data.append([x_noise/H, y_noise/H, 0, 0, run])

    # ============================
    # GUARDAR DOCUMENT
    # ============================

    df = pd.DataFrame(data, columns=["x", "y", "label", "sublabel", "set"])
    filename = f"test_dataset_{run}.csv"
    output_path = os.path.join(output_folder, filename)
    df.to_csv(output_path, index=False)

    print(f"Document creat: {output_path}")
```
